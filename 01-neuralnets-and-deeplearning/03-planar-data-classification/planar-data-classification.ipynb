## <b>Explore Dataset</b>
Let's explore the contrived non-linear dataset that will help illuminate the benefits of deep learning

We'll be using [this](https://colorhunt.co/palette/fbf5dda6cdc616404ddda853) colour-palette.

```python
COLOURS = [
  'DDA853', # orange
  '16404D', # navy
  'A6CDC6', # teal
  'FBF5DD' # beige
]
```

```python
X, Y = create_flower_data()
```

```python
Y = Y.ravel()

colors = ['#DDA853' if y == 0 else '#16404D' for y in Y]

scatter = go.Scatter(
  x=X[0, :],
  y=X[1, :],
  mode='markers',
  marker=dict(
    color=colors,
    size=5,
    line=dict(width=1, color='black')
  ),
  name='Data Points'
)

fig = go.Figure(data=[scatter])

fig.update_layout(
  title='Scatter Plot of Flower Dataset',
  xaxis_title='x1',
  yaxis_title='x2',
  width=800,
)

fig.show()
```

## <b>Simple Logistic Regression</b>
Let's see how a linear model would perform on a non-linear dataset like this one.

```python
# Train the logistic regression classifier
clf = sklearn.linear_model.LogisticRegressionCV();
clf.fit(X.T, Y.T);
```

```python
plot_decision_boundary(lambda x: clf.predict(x), X, Y)
```

```python
lr_prediction = clf.predict(X.T)
percentage = float((np.dot(Y,lr_prediction) + np.dot(1-Y,1-lr_prediction))/float(Y.size)*100)
print(f"Accuracy (% of correctly labelled points): {percentage}%")
```
